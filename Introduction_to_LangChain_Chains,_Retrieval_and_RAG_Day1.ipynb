## PromptTemplate

```python
from langchain_core.prompts import PromptTemplate

template1 = "Explain {topic} to a beginner. Answer only in one paragraph"

prompt1 = PromptTemplate.from_template(template1)

prompt1
```

```python
formatted_prompt1 = prompt1.format(topic = "RAG")
print(formatted_prompt1)
```

```python
template2 = "Write a short story set in {place} involving a character named {character} who has the goal of {goal}."

prompt2 = PromptTemplate.from_template(template2)

formatted_prompt2 = prompt2.format(
    place = "a haunted castle",
    character = "Alice",
    goal = "finding a hidden treasure"
)

formatted_prompt2
```

# Chain

```python
# LCEL (Langchain Expression Language)

chain = prompt1 | llm

result = chain.invoke({"topic": "Vector Databases"})
result
```

```python
result.content
```

```python
for each in result:
  print(each)
```

## Output Parser

```python
from langchain_core.output_parsers import StrOutputParser

parser = StrOutputParser()

chain = prompt1 | llm | parser

chain.invoke({"topic": "Vector Databases"})
```

```python
chain.invoke({"topic": "Embeddings"})
```

```python
# batch()

chain.batch([
    {"topic": "RAG"},
    {"topic": "Agents"},
    {"topic": "Embeddings"}
])
```

```python
chain.invoke({"topic": "Embeddings"})
```

```python
import time
for chunk in chain.stream({"topic": "Embeddings"}):
  print(chunk, end = "")
  time.sleep(0.1)
```

# Runnables

```python
from langchain_core.runnables import RunnableLambda


upper = RunnableLambda(
    lambda x:x.upper()
)

upper.invoke("hello world")
```

```python
chain = prompt1 | llm | parser | upper

response = chain.invoke({"topic": "Embeddings"})

print(response)
```

```python
first_few_tokens = RunnableLambda(
    lambda x: " ".join(x.split()[:20])
)

chain = prompt1 | llm | parser | upper | first_few_tokens

response = chain.invoke({"topic": "Embeddings"})

print(response)
```
